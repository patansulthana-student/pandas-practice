```python

import pandas as pd

data = {
    "Name": ["Asha", "Ravi", "Sita", "Rahul", "Priya", "John"],
    "Age": [20, 21, 19, 22, 20, 23],
    "Marks": [85, 90, 78, 88, 95, 82]
}

df = pd.DataFrame(data)

print("1. HEAD - First 5 rows")
display(df.head())

print("2. TAIL - Last 5 rows")
display(df.tail())

print("3. SHAPE - Number of rows and columns")
print(df.shape)

print("4. DESCRIBE - Statistical summary")
display(df.describe())
```

```python
#series object is one dimensional label array
import pandas as pd
s1=pd.Series([1,2,3,4,5])
s1
```

```python
#Changing Index
import pandas as pd
s1=pd.Series([1,2,3,4,5])
s1

import pandas as pd
s1=pd.Series([1,2,3,4,5],index=['a','b','c','d','e'])
s1
```

```python
#Series object from dictionary
import pandas as pd
pd.Series({'a':10,'b':20,'c':30})
```
