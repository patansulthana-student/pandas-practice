```python

import pandas as pd

data = {
    "Name": ["Asha", "Ravi", "Sita", "Rahul", "Priya", "John"],
    "Age": [20, 21, 19, 22, 20, 23],
    "Marks": [85, 90, 78, 88, 95, 82]
}

df = pd.DataFrame(data)

print("1. HEAD - First 5 rows")
display(df.head())

print("2. TAIL - Last 5 rows")
display(df.tail())

print("3. SHAPE - Number of rows and columns")
print(df.shape)

print("4. DESCRIBE - Statistical summary")
display(df.describe())
```

```python
#Line Plot
import numpy as np
from matplotlib import pyplot as plt

x = np.arange(1, 11)
x

y = 2 * x

plt.plot(x, y)
plt.show()
```

```python
#Adding Title and Labels
plt.plot(x,y)
plt.title("line plot")
plt.xlabel("x axis")
plt.ylabel("y axis")
```

```python
# Changing Line Aesthetics

plt.plot(x, y, color='g', linestyle=':', linewidth=2)
plt.show()
```

```python
#Adding two lines in the same plot
x=np.arange(1,11)
y1=2*x
y2=3*x
plt.plot(x, y1,color='g',linestyle=':',linewidth=2)
plt.plot(x, y2,color='r',linestyle='-.',linewidth=3)
plt.title("Line pLot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.grid(True)
plt.show()
```

```python
#Adding subplots
x=np.arange(1,11)
y1=2*x
y2=3*x
plt.subplot(1,2,1)
plt.plot(x, y1,color='r',linestyle=':',linewidth=2)
plt.subplot(1,2,2)
plt.plot(x, y2, color='g',linestyle=":",linewidth=2)
plt.subplot(2,1,1)
plt.plot(x, y1,color='r',linestyle=':',linewidth=2)
plt.subplot(2,1,2)
plt.plot(x, y1,color='g',linestyle=':',linewidth=2)
plt.show()
```
